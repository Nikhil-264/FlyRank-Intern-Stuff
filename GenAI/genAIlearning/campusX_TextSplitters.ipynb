# Text Splitting

Text splitting is the process of breaking large chunks of text(like articles, pdfs, books) into smaller, managable pieces that LLM can handle effectively

- Overcoming model limitations : Many embedding models and language models have maxumum input size constainsts. Splitting allows is to process documents that would otherwise exceed these limits

# 1. Length Based Text Splitting

size of chunks is decided (eg:100 characters)

```python
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader('Book1.pdf')

docs = loader.load()

splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=0,
    separator=""
)

result = splitter.split_documents(docs)

print(result[0])
```

- ChunkOverlap tells us that how much overlap between two chunks
- Like whatever the value we will set that much charcters will be in both the chunks

- Here dark green part is chunk overlap section

# 2. Text-Structure based

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter 

text = """
Cricket is more than just a sport in India; it is an emotion that connects millions of people across the country. From crowded stadiums to small streets where children play with tennis balls and plastic bats, cricket can be seen everywhere. The game is played between two teams of eleven players each, and the main objective is to score more runs than the opponent.

One of the most exciting formats of cricket is T20, where every match is fast-paced and full of entertainment. Leagues like the Indian Premier League (IPL) have changed modern cricket by bringing together international stars and young local talent. Legendary players such as Virat Kohli, MS Dhoni, and Sachin Tendulkar have inspired generations of fans with their performances and dedication.

Cricket also teaches important life lessons such as teamwork, patience, discipline, and leadership. A single match can change dramatically within a few overs, which is why fans say cricket is a game of glorious uncertainties. Whether it is a World Cup final or a friendly gully cricket match, the excitement and passion remain the same.

The popularity of cricket continues to grow because it combines skill, strategy, and emotion in a unique way. For many people, watching a cricket match with friends and family is not just entertainment — it is a celebration.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 0,
)

chunks = splitter.split_text(text)

print(len(chunks))
print(chunks)
```
