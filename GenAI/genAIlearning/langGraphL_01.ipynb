# What is Agentic AI

- Agentic AI is type of AI can take up a task or goal from a user and then work towards completing it on its own, with minimal human guidance
- It plans, takes action adapts to changes and seeks help only when necessaary

# Key Characteristics
1. Autonomous
2. Goal oriented
3. Planning
4. Reasoning
5. Adaptability
6. Context awareness

# 1. Autonomous

Autonomy refers to the AI system's ability to take actions on its own to achaieve a given goal, without needing step by step instructions

# 2. Goal Oriented
Being Goal oriented means that the AI system operates with a persistant objective in mind and continuously directs its actions to achieve that objective, rather than just responding to isolated propmts
1. Goals acts as a compass for autonomy
2. Goals can come with constrainsts
3. Goals are stored in core memory

4. Goals can be altered

# 4. Reasoning

Reasoning is the cognitive process through which an agentic AI systems interprets information draws conclusion and make dicisions, Both while planning ahead and while executing actions in real time

# 5. Adaptability

Adaptability is the agents ability to modify its plans, strategies or actions in response to unexpected conditions - all while staying alighned with the goal
1. Failueres(Calender API)
2. External feeback
3. Changing goals

# 6. Context Awareness

Context awareness is the agents ability to understand, retain and utilize relevant information from the ongoing task, past interations, user prefernces and environmental cues to make better decisions throughout a multistep process
