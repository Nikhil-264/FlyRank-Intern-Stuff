# State

In langGraph, state is the shared memory that flows through your workflow - it holds all the data being passed between nodes as your graph runs

at any node, node has access to state, and iis mutable, every node can make required changes in this state

# Reducers

Reducers in langgraph defines how updates from nodes are applied to the shared state

each key in the state can have its own reducer, whihc decides whether new data replaces, merges or adds to the existing value
