# Understanding RAG

- Indexing - External Knowledge base system making
- Retreival - understand the query and find the chunks related to query in the knowledge base
- Augmentation - You make prompt using query and retreived data 
- Generation - LLM generates the reposnse on all the prev steps

## 1. Indexing
Indexing is the process of preparing your knowledge base so that it can be efficiently searched at query time. This steps consists of 4 sub steps
- Document Ingestion
- Text Chunking
- Embeddings generation
- Storage in vector store

# 2. Retrieval

- Retireval is the real time process of finding the most relevant pieces of infromation from a pro built index(created during indexing) based in the users qeustion

# 3. Augmentation

Augmentation refers to the step where the retreived documents(chunks of relenva t context) are combined with the user's qeury to form a new enriched prompt for the LLM

# 4. Generation
Generation is the final step where a LLM uses the user's query and the retreived and augmented context to generate the response
