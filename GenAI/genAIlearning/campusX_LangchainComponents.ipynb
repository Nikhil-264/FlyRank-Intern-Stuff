## 5. Memory

LLM API calls are stateless


No context of previous input

this is solves using memory
Types:
1. conversationBufferMemory : stores a transcript of recent messages. Great for short chats but can grow large quickly

2. conversationBufferWindowMemory : Only keelps the last N interations to avois excessive token usage

3. SummerizerBasedMemory: Periodically summerizes older chats segments to keel=p a condensed memory footprint
