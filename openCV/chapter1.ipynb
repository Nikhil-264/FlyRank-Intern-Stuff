```python
cap = cv2.VideoCapture('Resources/Minion_Dance_Video_Generated.mp4')
while True:
    success, img = cap.read()
    if not success:
        break
    cv2.imshow("Video", img)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```
