### Part of Speech Tagging

## Applications of the POS Tagging
1. Named Entity Recognition
2. Question answering system
3. Word sensing disambiguation
4. Chatbots

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp(u"I will google about facebook")
doc.text
```

```python
doc[0]
```

```python
doc[0].pos_ # Course grade pos
```

```python
doc[3].tag_  # Fine grade pos
```

```python
spacy.explain('PRON')
```

```python
for word in doc:
    print(word.text, "---->", word.pos_, word.tag_, spacy.explain(word.tag_))
```

```python
doc3 = nlp(u"Before was was was, was was is")
for word in doc3:
    print(word.text, "---->", word.pos_, word.tag_, spacy.explain(word.tag_))
```

```python
doc4 = nlp(u"The quick brown fox jumps over the lazy dog")
```

```python
from spacy import displacy
displacy.render(doc4, style='dep', jupyter=True)
```

```python
options = {
    'distance' : 80,
    'compact' : True,
    'color' : '#fff',
    'bg' : '#00a65a'
}
```

```python
displacy.render(doc4, style='dep', jupyter=True, options=options)
```
