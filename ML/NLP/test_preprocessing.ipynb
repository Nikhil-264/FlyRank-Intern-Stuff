## 3. Remove URLs

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www.\S+')
    return pattern.sub(r'', text)
```

```python
text1 = 'Check out my notebook https://www.kaggle.com/campusx/notebook8223fc1abb'
text2 = 'Check out my notebook http://www.kaggle.com/campusx/notebook8223fc1abb'
text3 = 'Google search here www.google.com'
text4 = 'For notebook click https://www.kaggle.com/campusx/notebook8223fc1abb to search check www.google.com'
```

```python
remove_url(text1)
remove_url(text2)
remove_url(text3)
remove_url(text4)
```

## 4. Remove Punctuations

```python
import string, time
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punctuations(text):
    for char in exclude:
        text = text.replace(char, '')
    return text
```

```python
text = 'String, with .;? pucntuations'
```

```python
start = time.time()
print(remove_punctuations(text))
time1 = time.time() - start
print(time1 * 50000)
```

```python
def remove_punc1(text):
    return text.translate(str.maketrans('', '', exclude))
```

```python
start = time.time()
remove_punc1(text)
time2 = time.time() - start
```

## 7. Removing StopWords

```python
# a, the, are, of, my etc
# Stop words are removed unless we are doing the Parts od Speech tagging
```

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []
    
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
text5 = 'I am the best footballer in the world and im gonna be even better than me tomorrow than my past self'
```

```python
remove_stopwords(text5)
```

## 8. Handling Emojis

```python
# Either you remove them
# Or replace it with the meaning of that emoji
```

```python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                                u"\U0001F600-\U0001F64F"
                                u"\U0001F300-\U0001F5FF"
                                u"\U0001F680-\U0001F6FF"
                                u"\U0001F1E0-\U0001F1FF"
                                u"\U00002702-\U000027B0"
                                u"\U000024C2-\U0001F251"
                                "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

## 10. Stemming

In grammar, inflection is the modification of a word to express different grammatical categories such as tense, case, voice, aspect, person, gender and mood

Stemming the process of reducing inflection in words to their root forms such as mapping a group of words to the same stem even is the stem itself is not a valid word in the language

Used in information retreival systems

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = 'walk walked walking walks probability'
stem_words(sample)
```

## 11. Lemmetization

Lemmatization, unlike Stemming, reduces the inflected words properly ensuring that the root word belongs to the language. In Lemmatization root word is called Lemma. A lemma (plural lemmas or lemmata) is the canonical form, dictionary form, or citation form of a set of words.

```python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at the same time."
punctuation = "?:!.,;"
sentence_words = nltk.word_tokenize(sentence)

for word in sentence_words:
    if word in punctuation:
        sentence_words.remove(word)
        
sentence_words
print("{0:20} {1:20}".format("Word", "Lemma"))
for word in sentence_words:
    print("{0:20} {1:20}".format(word, wordnet_lemmatizer.lemmatize(word)))
```
