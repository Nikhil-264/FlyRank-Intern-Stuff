```python
a = np.array([[1, 2, 3], [4, 5, 6]])
zeros_like_a = np.zeros_like(a)  # array with same shape as 'a' filled with zeros
ones_like_a = np.ones_like(a)    # array with same shape as 'a' filled with ones
full_like_a = np.full_like(a, 7) # array with same shape as 'a' filled with 7
```

```python
print(a)
```

```python
print(zeros_like_a)
```

```python
print(ones_like_a)
```

```python
print(full_like_a)
```
